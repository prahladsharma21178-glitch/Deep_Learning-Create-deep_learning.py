***problrm: build one model that can reach from 0 to 5
states= 0 to 5
start: 0
goal=5

actions:
0: if model move left
1: if model move right

reward:
reaching goal: +10
other steps :   -1

```python
import numpy as np
import random
```

```python
# initial state and action
states=6
actions=2
# initial Quality matrix
Q=np.zeros((states,actions))
Q
```

```python
# Q learning parameters
alpha=0.1
gamma=0.9
epsilon=0.2
episodes=500
```

```python
# to train the model for 500 time:
for _ in range(episodes):
  state=0
  while state!=5:
    #epsilon greedy approach
    if random.uniform(0,1)<epsilon:
      actions=random.choice([0,1])
    else:
      actions=np.argmax(Q[state])
    next_state= state-1 if actions==0  else state+1
    # set the limit of model 0 to 5
    next_state=max(0,min(next_state,5))

    # rewards
    reward =10 if next_state==5 else -1
    Q[state,actions]=Q[state,actions]+alpha*(reward+gamma*np.max(Q[next_state]-Q[state,actions]))
    state=next_state
```
